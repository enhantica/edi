# Y₂O₃ [Ia-3] on SPODI @ FRM II

```python
import edi

project = edi.Project.load('pd-neut-cwl_y2o3-spodi')
```

```python
result = project.analysis.fit()
print(f'Rwp: {100 * result.rwp:.2f}%   ·   Reduced χ²: {result.reduced_chi_square:.2f}')
```

```python
import plotly.graph_objects as go

data = project.experiments[0].data
figure = go.Figure([
    go.Scatter(
        x=data.two_theta,
        y=data.intensity_meas,
        name='Measured',
        mode='lines',
        line={'color': '#0084c7', 'width': 1.5},
    ),
    go.Scatter(
        x=data.two_theta,
        y=data.intensity_calc,
        name='Calculated',
        mode='lines',
        line={'color': '#d13f73', 'width': 1},
    ),
])
figure.update_layout(
    template='plotly_white',
    height=340,
    margin={'l': 65, 'r': 25, 't': 25, 'b': 50},
    xaxis_title='2θ (°)',
    yaxis_title='Intensity',
    legend={'orientation': 'h', 'y': 1.12},
)
figure.show(renderer='plotly_mimetype')
```
